```python
import arakawa as ar
```

Control Blocks are for composing an HTML form with [Compute](../../reference/layout-blocks/compute/) layout block.

Arakawa supports various control blocks.

```python
ar.TextBox("text")
```

```python
ar.NumberBox("number")
```

```python
ar.RangeField(name="range", min=0, max=100, step=10)
```

```python
ar.SwitchField("switch")
```

```python
ar.FileField("file")
```

```python
ar.DateTimeField("date-time")
```
